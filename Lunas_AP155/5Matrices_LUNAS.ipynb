### Problem 1: (Circuit Analysis)


Consider a long chain of resistors wired up like this:

<img src="reschain.png" alt="Alt text" width=95%>

All the resistors have the same resistance~$R$.  The power rail at the top is at voltage~$V_+=5$V.  The problem is to find the voltages $V_1...V_N$ at the internal points in the circuit.

1. **Using Ohm's law and the Kirchhoff current law**, which says that the total net current flow out of (or into) any junction in a circuit must be zero, show that the voltages $V_1\ldots V_N$ satisfy the equations

$$3V_1 - V_2 - V_3 = V_+ $$
$$-V_1 + 4V_2 - V_3 - V_4 = V_+ $$
      
$$-V_{i-2} - V_{i-1} + 4V_i - V_{i+1} - V_{i+2} = 0 $$

$$-V_{N-3} - V_{N-2} + 4V_{N-1} - V_N = V_- = 0 $$
$$-V_{N-2} - V_{N-1} + 3V_N = V_- = 0 $$

Note: The assignment of $V_-$ is to make the equations symmetric

2. Express these equations in vector form~$A\vec{v} = \vec{w}$ and find the values of the matrix~$A$ and the vector~$\vec{w}$.

3. Write a program to solve for the values of the~$V_i$ when there are $N=6$ internal junctions with unknown voltages.  (Hint: All the values of $V_i$ should lie between zero and $5$V.  If they don't, something is wrong.)

4. Now repeat your calculation for the case where there are $N=10\,000$ internal junctions.  This part is not possible using standard tools like the *solve* function.  You need to make use of the fact that the matrix~$\vec{A}$ is banded.  

### 1.

According to **Kirchhoff's Current Law (KCL)**, the total net current flowing out of junction $V_i$ must equal zero:
$$\sum I_{\text{out}} = 0$$

By **Ohm's Law**, current flowing through a resistor of resistance $R$ from node $i$ to a adjacent node $j$ is:
$$I = \frac{V_i - V_j}{R}$$

For a general internal node $V_i$, it is connected via identical resistors $R$ to four neighbors: $V_{i-2}$, $V_{i-1}$, $V_{i+1}$, and $V_{i+2}$. Setting the sum of currents to zero:

$$\frac{V_i - V_{i-2}}{R} + \frac{V_i - V_{i-1}}{R} + \frac{V_i - V_{i+1}}{R} + \frac{V_i - V_{i+2}}{R} = 0$$

Multiplying the entire equation by $R$:

$$(V_i - V_{i-2}) + (V_i - V_{i-1}) + (V_i - V_{i+1}) + (V_i - V_{i+2}) = 0$$

Combining the $V_i$ terms ($V_i + V_i + V_i + V_i = 4V_i$) and rearranging in index order gives:

$$-V_{i-2} - V_{i-1} + 4V_i - V_{i+1} - V_{i+2} = 0$$

### 2. 

The linear system of equations can be expressed in matrix form $\vec{A}\vec{v} = \vec{w}$, where $\vec{v} = [V_1, V_2, \dots, V_N]^T$ is the vector of unknown junction voltages.

* **Matrix $\vec{A}$ ($N \times N$ pentadiagonal matrix):**

$$\vec{A} = \begin{bmatrix}
 3 & -1 & -1 &  0 &  0 & \dots &  0 &  0 \\
-1 &  4 & -1 & -1 &  0 & \dots &  0 &  0 \\
-1 & -1 &  4 & -1 & -1 & \dots &  0 &  0 \\
 0 & -1 & -1 &  4 & -1 & \dots &  0 &  0 \\
\vdots & & \ddots & \ddots & \ddots & \ddots & & \vdots \\
 0 &  0 & \dots & -1 & -1 &  4 & -1 & -1 \\
 0 &  0 & \dots &  0 & -1 & -1 &  4 & -1 \\
 0 &  0 & \dots &  0 &  0 & -1 & -1 &  3
\end{bmatrix}$$

* **Vector $\vec{w}$ ($N \times 1$ vector):**

$$\vec{w} = \begin{bmatrix} V_+ \\ V_+ \\ 0 \\ \vdots \\ 0 \\ 0 \end{bmatrix} = \begin{bmatrix} 5 \\ 5 \\ 0 \\ \vdots \\ 0 \\ 0 \end{bmatrix}$$


```python
import numpy as np
import matplotlib.pyplot as plt
import math

# Setup parameters
N = 6
V_plus = 5.0  # Volts

# Build matrix A
A = np.zeros((N, N))
for i in range(N):
    A[i, i] = 4
    if i - 1 >= 0: A[i, i - 1] = -1
    if i - 2 >= 0: A[i, i - 2] = -1
    if i + 1 < N: A[i, i + 1] = -1
    if i + 2 < N: A[i, i + 2] = -1

# Boundary conditions
A[0, 0] = 3
A[-1, -1] = 3

# Build vector w
w = np.zeros(N)
w[0] = V_plus
w[1] = V_plus

# Solve A * v = w
v = np.linalg.solve(A, w)

# Display results
print(f"Voltage solutions for N = {N}:")
for i, voltage in enumerate(v, start=1):
    print(f"V_{i} = {voltage:.4f} V")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math
from scipy.linalg import solve_banded

# Setup parameters
N = 10000
V_plus = 5.0  # Volts

# Matrix A is pentadiagonal (2 diagonals below, 2 diagonals above)
# solve_banded expects array of shape (l + u + 1, N) -> (5, N)
ab = np.zeros((5, N))

ab[0, 2:] = -1   # 2nd super-diagonal
ab[1, 1:] = -1   # 1st super-diagonal
ab[2, :]  =  4   # Main diagonal
ab[3, :-1] = -1  # 1st sub-diagonal
ab[4, :-2] = -1  # 2nd sub-diagonal

# Boundary conditions
ab[2, 0] = 3     # A[0, 0]
ab[2, -1] = 3    # A[N-1, N-1]

# Build vector w
w = np.zeros(N)
w[0] = V_plus
w[1] = V_plus

# Solve linear system
v_10k = solve_banded((2, 2), ab, w)

# Output summary results
print(f"Successfully solved system for N = {N}:")
print(f"First 5 voltages: {v_10k[:5]}")
print(f"Last 5 voltages:  {v_10k[-5:]}")
print(f"All voltages between 0V and 5V: {np.all((v_10k >= 0) & (v_10k <= V_plus))}")

# Plot voltage distribution along the chain
plt.figure(figsize=(10, 5))
plt.plot(np.arange(1, N + 1), v_10k, color='blue', linewidth=1.5)
plt.title(f"Voltage Profile across N = {N} Junctions")
plt.xlabel("Junction Index (i)")
plt.ylabel("Voltage (V)")
plt.grid(True)
plt.show()
```

### Some codes for solving matrix problems

* solving matrix equations ($\bf{A}\vec{x} = \vec{b}$) -> `np.linalg.solve`
* LU-decomposition ($\bf{A} = \bf{L}\bf{U}$) -> `scipy.linalg.lu`
* matrix inversion ($\bf{A}^{-1}$) -> `np.linalg.inv`
* QR-decomposition ($\bf{A} = \bf{Q}\bf{R}$) -> `scipy.linalg.qr`
* eigenvalue problem ($\bf{A}\vec{x} = \lambda \vec{x}$) -> `np.linalg.eig`

Solving matrix problems by themselves isn't difficult. There is an abundance of linear algebra libraries that are ready to use: from the battle-tested ones (the classic [BLAS](https://www.netlib.org/blas/) and [LAPACK](https://www.netlib.org/lapack/)), HPC tools for large problems ([ScaLAPACK](https://netlib.org/scalapack/)), to hardware-accelerated ones often with GPUs ([cuBLAS](https://docs.nvidia.com/cuda/cublas/index.html) and [cuSolver](https://docs.nvidia.com/cuda/cusolver/index.html) for NVIDIA). 

The hardest part is actually composing the matrix. How do you transform a physical problem into a linear algebra problem?
